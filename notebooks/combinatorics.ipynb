# Lab 1 - Combinatorics, Microstates, and Thermodynamics

Runs the combinatorics code in `src/combinatorics.py`, its unit tests, and generates high-resolution vector figures for the report.

## Setup
Adds the repository root directory to Python's `sys.path` so that modules inside `src/` can be imported seamlessly.

```python
import os
import sys
import subprocess
from pathlib import Path

ROOT = Path.cwd() if (Path.cwd() / "src").exists() else Path.cwd().parent
sys.path.insert(0, str(ROOT))

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.backends.backend_pdf import PdfPages

from src.combinatorics import (
    omega_exact,
    ln_omega_logspace,
    omega_logspace,
    ln_stirling,
    ln_omega_stirling,
    ln_prob_asymptotic,
    plot_probability_distribution_fraction,
    plot_probability_distributions,
    plot_multiplicity
)

print("Repo root:", ROOT)
```

## Unit Tests
Executes `pytest` on the `tests/` directory to verify all test cases pass successfully.

```python
result = subprocess.run([sys.executable, "-m", "pytest", "-v"], cwd=ROOT, capture_output=True, text=True)
print(result.stdout if result.stdout else result.stderr)
```

## Exact Multiplicity Breakdown for $N = 7$

```python
N = 7
print(f"{'macrostate (k heads)':<22}{'Omega(N, k)':>14}")
print("-" * 38)
for k in range(N + 1):
    omega_val = omega_exact(N, k)
    print(f"{k heads':<22}{omega_val:>14}")
print("-" * 38)
print(f"{'Total sum (2^N)':<22}{sum(omega_exact(N, k) for k in range(N + 1)):>14}")
```

## Log-Space Multiplicity & Stirling Approximation Analysis

```python
N_test = 100
k_test = 50
print(f"N = {N_test}, k = {k_test}:")
print(f"  Exact/Log-space Omega: {omega_logspace(N_test, k_test):.4e}")
print(f"  Stirling ln(Omega):    {ln_omega_stirling(N_test, k_test):.4f}")
print(f"  Asymptotic ln P(k):    {ln_prob_asymptotic(N_test, k_test):.4f}")
```

## Macrostate Multiplicity and Probability Distributions
Visualizing multiplicity in logarithmic scale and probability distributions in linear and fraction space for $N \in \{20, 50, 80, 100\}$[cite: 1].

```python
fig1 = plot_multiplicity()
plt.show()

fig2 = plot_probability_distributions()
plt.show()

fig3 = plot_probability_distribution_fraction()
plt.show()
```

## Exporting Figures to PDF
Saves all generated figures into vector PDF format inside `report/figures/` as required by the guidelines[cite: 2].

```python
output_dir = ROOT / "report" / "figures"
output_dir.mkdir(parents=True, exist_ok=True)
output_path = output_dir / "combined_figures.pdf"

with PdfPages(output_path) as pdf:
    fig1 = plot_multiplicity()
    pdf.savefig(fig1)
    plt.close(fig1)
    
    fig2 = plot_probability_distributions()
    pdf.savefig(fig2)
    plt.close(fig2)
    
    fig3 = plot_probability_distribution_fraction()
    pdf.savefig(fig3)
    plt.close(fig3)

print(f"Successfully saved figures to: {output_path}")
```
